# Exercici 9
Completeu la funció absents_reals, que ha de marcar com a absent qualsevol dels tres graus
d’absència:

PSEUDO = {...} # els pseudo-buits que heu trobat (després de treure espais)

def absents_reals(s, sentinelles=()):

m = ... # NaN honestos

m = m | ... # pseudo-buits: passeu a text i traieu els espais

m = m | ... # sentinelles numèriques

return m

Feu-la servir per construir una taula amb, per a cada columna, el % de NaN, de pseudo-buits, de
sentinelles i el % total d’absents reals. Després, calculeu la mitjana de km tal com és i excloent-ne
les sentinelles, i expliqueu la diferència.

```python
PSEUDO = ['', '-', 'desconegut', 'no en té']
```

```python
def absents_reals(s, sentinelles=()):

    # NaN honestos
    m = s.isna()

    # Pseudo-buits: convertir a texto y eliminar espacios
    m = m | s.astype('string').str.strip().str.lower().isin(PSEUDO)

    # Sentinelles numèriques
    m = m | s.isin(sentinelles)

    return m
```

```python
sentinelles = [-1, -999, 9999]
```

```python
resultats = []

for columna in df.columns:

    s = df[columna]

    # Porcentaje de NaN
    nan = s.isna().mean() * 100

    # Porcentaje de pseudo-buits
    pseudo = (
        s.astype('string')
        .str.strip()
        .str.lower()
        .isin(PSEUDO)
        .mean() * 100
    )

    # Porcentaje de sentinelles
    sent = s.isin(sentinelles).mean() * 100

    # Porcentaje total de ausentes reales
    total = absents_reals(s, sentinelles).mean() * 100

    resultats.append({
        'columna': columna,
        'NaN (%)': nan,
        'Pseudo-buits (%)': pseudo,
        'Sentinelles (%)': sent,
        'Total absents (%)': total
    })

taula_absents = pd.DataFrame(resultats)

print(taula_absents.round(2).T)
```

```python
# Media original
media_original = df['km'].mean()

# Media excluyendo los sentinelles
media_sin_sentinelles = df.loc[
    ~df['km'].isin(sentinelles), 'km'
].mean()

print("Media original:", media_original)
print("Media sin sentinelles:", media_sin_sentinelles)
```

# Exercici 11
Poseu un exemple plausible, en aquest negoci de trasllats, d’una variable on l’absència podria ser
MNAR. Expliqueu per què aquest patró no es pot demostrar només amb les dades que teniu i
quin risc tindria omplir-ne els forats amb la mitjana.

### Exemple de variable MNAR: `km` en els trasllats llargs

**Exemple plausible.** En els trasllats molt llargs (per exemple, Palma–Cala Rajada o un trajecte cap a Formentor), l'empresa sovint pacta un **preu tancat** amb el client. Com que el preu no depèn dels quilòmetres, el conductor o l'operador no s'hi fixa i **no anota els km**. L'absència de `km` dependria, doncs, **del mateix valor que falta**: com més llarg és el trasllat, més probable és que no tinguem els km. Això és MNAR (*Missing Not At Random*).

**Per què no es pot demostrar amb les dades que tenim.** Per saber si falten sobretot els km llargs hauríem de conèixer els km que falten, i precisament no els tenim. Les dades observades només ens mostren els trasllats que sí que tenen km, i aquests semblen normals. Amb les dades disponibles, un patró MNAR no es distingeix d'un MCAR o d'un MAR. Com a molt, podríem trobar pistes indirectes (per exemple, que les reserves sense km tinguin imports més alts), però mai una prova. Per confirmar-ho caldria una font externa: el GPS dels vehicles, les factures o preguntar al negoci com es registren els trasllats amb preu tancat.

**Risc d'omplir els forats amb la mitjana.** La mitjana de `km` (uns 24,5 km sense sentinelles) es calcula només amb els trasllats que tenen km, que són sobretot els curts. Si omplim amb aquest valor uns trasllats que en realitat eren llargs:
- **Esbiaixem les dades a la baixa:** la mitjana i el total de km quedarien per sota de la realitat.
- **Reduïm artificialment la variabilitat:** la desviació típica baixaria i desapareixerien els trasllats llargs legítims, que són justament els més interessants.
- **Creem inconsistències noves:** una reserva amb un import alt i 24,5 km imputats tindria un preu per km absurd i semblaria un error.
- **Amaguem el problema:** després de la imputació ja no es veu que faltaven dades, i les decisions de negoci (tarifes, planificació de vehicles, consum de combustible) es prendrien sobre dades falses.

# Exercici 13
Calculeu, amb expressions regulars (str.match), el % de valors de creada_el en format
aaaa-mm-dd i en format dd/mm/aaaa. Proveu de convertir la columna amb pd.to_datetime(...,
format="mixed", dayfirst=True) i comproveu què passa amb una data ISO com 2026-03-05.
Per què l'ajuda de l'exercici del validador converteix cada format per separat?

```python
creada_txt = df["creada_el"]

es_iso = creada_txt.str.match(r"^\d{4}-\d{2}-\d{2}$")      # aaaa-mm-dd
es_europea = creada_txt.str.match(r"^\d{2}/\d{2}/\d{4}$")  # dd/mm/aaaa

print(f"Format aaaa-mm-dd: {es_iso.sum():4d} ({es_iso.mean()*100:5.2f} %)")
print(f"Format dd/mm/aaaa: {es_europea.sum():4d} ({es_europea.mean()*100:5.2f} %)")
print(f"Cap dels dos:      {(~es_iso & ~es_europea).sum():4d}")
```

```python
# 1) Què passa amb una data ISO si fem servir format="mixed" i dayfirst=True?
prova = pd.to_datetime(pd.Series(["2026-03-05"]), format="mixed", dayfirst=True)
print("2026-03-05 ->", prova[0].date(), "(hauria de ser 5 de març)")
print()

# 2) Convertim tota la columna de les dues maneres i comparem
mixed = pd.to_datetime(creada_txt, format="mixed", dayfirst=True)

per_separat = pd.to_datetime(creada_txt, format="%Y-%m-%d", errors="coerce").fillna(
    pd.to_datetime(creada_txt, format="%d/%m/%Y", errors="coerce")
)

diferents = mixed != per_separat
print("Dates diferents entre els dos mètodes:", diferents.sum())
print("D'aquestes, quantes eren ISO:", (diferents & es_iso).sum())
print()
print(pd.DataFrame({
    "text_original": creada_txt[diferents],
    "amb_mixed": mixed[diferents].dt.date,
    "per_separat": per_separat[diferents].dt.date,
}).head())
```

**Formats de `creada_el`:** 757 valors (93,11 %) estan en format `aaaa-mm-dd` i 56 (6,89 %) en format `dd/mm/aaaa`. No n'hi ha cap d'un altre format, així que la columna barreja exactament dos formats (**mala codificació**).

**Què passa amb `format="mixed", dayfirst=True`:** la data ISO `2026-03-05` (5 de març) es converteix en **2026-05-03 (3 de maig)**. `dayfirst=True` fa que pandas interpreti el primer número després de l'any com el dia, també en les dates ISO, i intercanvia el dia i el mes. A tota la columna passa en **263 dates**, totes ISO: les que tenen el dia entre 1 i 12, perquè llavors el dia també és un mes vàlid. Les dates amb el dia més gran que 12 (per exemple `2026-04-23`) no es poden intercanviar i es converteixen bé.

El més perillós és que **no dona cap error ni cap avís**: la conversió sembla correcta però una tercera part de les dates són falses, i després fallarien regles com `recollida_abans_creada` o `creada_futura`.

**Per què el validador converteix cada format per separat:** perquè cada format té un ordre de dia i mes diferent, i un sol criteri (`dayfirst` sí o no) sempre n'espatlla un dels dos. Fent servir un `format=` explícit per a cada un (`%Y-%m-%d` i `%d/%m/%Y`) amb `errors="coerce"`, cada conversió només accepta les dates del seu format i deixa les altres com a `NaT`. Després, `fillna` combina els dos resultats. Així no hi ha cap ambigüitat i, si alguna data no encaixés en cap dels dos formats, quedaria com a `NaT` i la veuríem, en lloc d'amagar-se amb un valor equivocat.
